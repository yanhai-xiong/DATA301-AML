```python
import string

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader

import matplotlib.pyplot as plt
import numpy as np
import torch.nn.functional as F
```

```python
# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
# Define a transform to normalize the data
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])
```

```python
# Download and load the training and test datasets
train_dataset = torchvision.datasets.EMNIST(root='./data', train=True, split='letters', transform=transform, download=True)
test_dataset = torchvision.datasets.EMNIST(root='./data', train=False, split='letters', transform=transform, download=True)

train_loader = DataLoader(dataset=train_dataset, batch_size=64, shuffle=True)
# you can specify batsh size for the test if you plan to check the results at each time of the training process
test_loader = DataLoader(dataset=test_dataset,shuffle=False)
```

```python
for X, y in train_loader:
    print(f"Shape of X [N, C, H, W]: {X.shape}")
    print(f"Shape of y: {y.shape} {y.dtype}")
    break
# Define the class names
class_names = list(string.ascii_lowercase)
```

```python
# look at some sample images
i = 13
plt.imshow(X[i,0,:].T,cmap='viridis_r')
plt.show()
print(class_names[y[i].item()-1])
```

```python
# Function to display images
def imshow(img, label):
    img = img / 2 + 0.5  # unnormalize
    npimg = img.numpy()
    plt.imshow(npimg[0].T,cmap='viridis_r')
    plt.title('Sample of Fashion/Apparel Items')
    #plt.title(f'First Item: {class_names[label]}')
    plt.show()

# Get some random training images
dataiter = iter(train_loader)
images, labels = next(dataiter)

# Show images
imshow(torchvision.utils.make_grid(images), labels[0])

# Display the name of the label for the first image in the batch
print(f'First Label: {class_names[labels[0]-1]}')
```

```python
# Define the convolutional model
class CNN(nn.Module):
    def __init__(self):
        super(CNN, self).__init__()
        self.layer1 = nn.Sequential(
            nn.Conv2d(1, 100, kernel_size=3, padding=1),
            nn.BatchNorm2d(100), # this is the z-score StandardScaler
            nn.ReLU(), # activations
            nn.MaxPool2d(kernel_size=2, stride=2), # summarizations to address the enlarged memory footprint
            nn.Dropout(p=0.2) # cutting dendrites randomly to prevent dead neurons
        )
        self.layer2 = nn.Sequential(
            nn.Conv2d(100, 100, kernel_size=3),
            nn.BatchNorm2d(100),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2,stride=2),
            nn.Dropout(p=0.2)
        )
        self.fc1 = nn.Linear(100*6*6, 1000) # here we pass on the 100 features extracted and each feature is now 6x6
        self.fc2 = nn.Linear(1000, 26) # is non-negotiable

    def forward(self, x):
        out = self.layer1(x)
        out = self.layer2(out) # here we got all the 100 features
        out = out.view(out.size(0), -1)
        out = self.fc1(out)
        out = self.fc2(out)
        return out

model = CNN().to(device)

# Define the loss function and optimizer
loss_fn = nn.CrossEntropyLoss() # it's a must: choose the correct loss function
optimizer = optim.Adam(model.parameters(), lr=0.001) # we chose adaptive momentum gradient descent as our optimizer

# Train the model
num_epochs = 10
```

```python
# here we define the training procedure
def train(dataloader, model, loss_fn, optimizer):
    size = len(dataloader.dataset)
    model.train()
    for batch, (X, y) in enumerate(dataloader): # this is the must-have iterator
        X, y = X.to(device), y.to(device)
        y = y - 1
        # Compute prediction error
        pred = model(X)
        loss = loss_fn(pred, y)

        # Backpropagation
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()

        if batch % 50 == 0:
            loss, current = loss.item(), (batch + 1)*len(X)
            print(f"loss: {loss:>7f}  [{current:>5d}/{size:>5d}]")
```

```python
# to evaluate the trained model on the Test data
def test(dataloader, model, loss_fn):
    size = len(dataloader.dataset)
    num_batches = len(dataloader)
    model.eval() # we set the model in evaluation state
    test_loss, correct = 0, 0
    with torch.no_grad(): # a must-have for model.eval()
        for X, y in dataloader:
            y = y - 1
            X, y = X.to(device), y.to(device)
            pred = model(X)
            test_loss += loss_fn(pred, y).item()
            correct += (pred.argmax(1) == y).type(torch.float).sum().item()
    test_loss /= num_batches
    correct /= size
    print(f"Test Error: \n Accuracy: {(100*correct):>0.1f}%, Avg loss: {test_loss:>8f} \n")
```

```python
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
```

```python
epochs = 10
for t in range(epochs):
    print(f"Epoch {t+1}\n-------------------------------")
    train(train_loader, model, loss_fn, optimizer)
    test(test_loader, model, loss_fn)
print("Done!")
```
