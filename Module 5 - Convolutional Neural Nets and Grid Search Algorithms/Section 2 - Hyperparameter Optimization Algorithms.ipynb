## Grid Search vs. Random Search

**Goal:** Find the hyperparameter configuration that minimizes validation loss:
$$
\lambda^*=\arg\min_{\lambda\in\Lambda}\mathcal{L}_{val}(\lambda)
$$

For example, for an MLP, hyperparameters might include the **number of hidden layers**, **neurons per layer**, **learning rate**, and **dropout rate**.

1. **Grid Search** systematically evaluates **every combination** of predefined hyperparameter values.
    - **Advantages:** Simple, reproducible, and exhaustive over the specified grid.
    - **Limitations:** Computational cost grows rapidly as more hyperparameters and candidate values are added.
2. **Random Search** samples hyperparameter configurations randomly from specified distributions or sets, instead of evaluating every combination.
    - **Advantages:** Flexible computational budget; can explore broader search spaces and continuous distributions.
    - **Limitations:** Does not use results from previous evaluations to guide subsequent choices.

We can use algorithms that do not require the gradient of the objective function in order to estimate the value for the *hyperparameters* and thus complete the *model selection* process.


The simplest *grid search algorithm* is an exhaustive search, namely loop through all the candidates and pick the one that is the *best* according to predefined criteria, such as minimizing the k-fold cross-validated loss or maximizing the accuracy, etc.

## <font color='navy' size=6pt> Swarm Algorithms - Particle Swarm Optimization</font>

Main Idea: consider a collection of particles that move around the search space influenced by their own best past location and the best past location of the whole swarm or a close neighbor.


<figure>
<center>
<img src='https://i.imgur.com/WOsFccl.png'
width='500px' />
</center>
</figure>

### PSO Formulation
The particle positions get updated: $$\large p_{i} (t+1)=p_i  (t)+v_i (t+1)$$

where velocity is 
$$\large v_i (t+1)=w\cdot v_i (t)+c_1⋅\chi_1⋅(p_i^{\text{best}} (t)−p_i (t))+c_2⋅\chi_2⋅(p^{\text{gbest}} (t)−p_i (t))$$
- $w$: **Inertia weight**, controlling how much of the previous velocity is retained.
- $c_1$: **Cognitive coefficient**, controlling attraction toward the particle's personal best.
- $c_2$: **Social coefficient**, controlling attraction toward the swarm's global best.
- $\chi_1,\chi_2 \sim\text{Uniform}(0,1)$: Independent random factors (typically sampled separately for each particle and dimension).

### Interpretation 

- **Larger $w$:** Maintains momentum and encourages exploration.
- **Smaller $w$:** Reduces momentum, often encouraging local refinement.
- **Larger $c_1$:** Emphasizes individual exploration around personal best positions.
- **Larger $c_2$:** Encourages particles to converge toward the swarm's best-known position.

### **Practical Heuristics**

- **Swarm size:** Start with approximately **20–40 particles**, adjusting based on problem complexity and evaluation cost.
- **Inertia weight ($w$):** Often chosen around $0.4$–$0.9$; it may decrease over iterations to shift from exploration toward exploitation.
- **Learning coefficients ($c_1,c_2$):** A common starting point is $c_1=c_2=2$, although the best values depend on the problem and stability considerations.
- **Velocity limits:** Optionally restrict maximum velocity to a fraction of each dimension's search range to prevent excessively large movements.
- **Boundary handling:** Keep particles within the feasible search space using clipping, reflection, or other constraint-handling strategies.
- **Neighborhood topology:** Instead of using the entire swarm's global best, particles can follow a **local neighborhood best** to maintain diversity and reduce premature convergence.

**Connection to Machine Learning:** Each particle can represent a candidate hyperparameter configuration, such as $(\text{learning rate},\text{dropout rate},\text{number of neurons})$. The objective is to find the configuration with the lowest validation loss.

*Note:* Standard PSO operates in continuous search spaces. Integer or categorical hyperparameters (e.g., number of layers or activation function) require discrete encodings or adapted PSO variants.

#### PSO - Implementation Idea

Create a data structure (e.g., Python dictionary) that, for each particle, records the position, velocity and the particle's best position from the start.

````python
# PSO algorithm

def pso(obj_function, num_dimensions, num_particles, max_iter,i_min=-10,i_max=10,bounds=None,w=0.5,c1=0.3,c2=0.6):
    # Initialize the particles
    # This creates a data structure such as a dictionary
    # very important: initiate a tracker variables
    if bounds is None:
        particles = [({'position': [np.random.uniform(i_min, i_max) for _ in range(num_dimensions)],
                    'velocity': [np.random.uniform(-1, 1) for _ in range(num_dimensions)],
                    'pbest': float('inf'),
                    'pbest_position': None})
                    for _ in range(num_particles)]
    else:
        particles = [({'position': [np.random.uniform(bounds[i][0], bounds[i][1]) for i in range(num_dimensions)],
                    'velocity': [np.random.uniform(-1, 1) for _ in range(num_dimensions)],
                    'pbest': float('inf'),
                    'pbest_position': None})
                    for _ in range(num_particles)]

    history = {
        'positions': [],
        'gbest_values': [],
        'gbest_positions': []
    }
    # Initialize global best
    gbest_value = float('inf')
    gbest_position = None

    for _ in range(max_iter):
        for particle in particles:
            position = particle['position']
            velocity = particle['velocity']

            # Calculate the current value
            current_value = obj_function(position[0],position[1])

            # Update personal best
            if current_value < particle['pbest']:
                particle['pbest'] = current_value
                particle['pbest_position'] = position.copy()

            # Update global best
            if current_value < gbest_value:
                gbest_value = current_value
                gbest_position = position.copy()

            # Update particle's velocity and position
            for i in range(num_dimensions):
                r1, r2 = np.random.uniform(), np.random.uniform()
                # here below we have the equation from the slides
                velocity[i] = w * velocity[i] + c1*r1 * (particle['pbest_position'][i] - position[i]) + c2*r2 * (gbest_position[i] - position[i])
                position[i] += velocity[i]
                # legalize the values to the provided bounds
                if bounds is not None:
                    position[i] = np.clip(position[i],bounds[i][0],bounds[i][1])

        history['positions'].append(
            np.array([p['position'].copy() for p in particles])
        )
        history['gbest_values'].append(gbest_value)
        history['gbest_positions'].append(gbest_position.copy())
    return gbest_position, gbest_value, history
````

## <font color='navy' size=6pt> Simulated Annealing</font>

**Main Idea**: drop a random trial point and then, based on the value of the objective function, continue to drop other points at a certain distance. The distance between consecutive points is measured as a function of the "temperature" which is decreasing based on an exponential law.

<figure>
<center>
<img src='https://spotintelligence.com/wp-content/uploads/2025/10/simulated-annealing.jpg'
width='500px' />
</center>
</figure>

### SA Key Idea

**Simulated Annealing** is an optimization algorithm that explores the search space using **one candidate solution at a time**. It occasionally accepts worse solutions to escape local minima.
1. **Start** with a random solution and a high temperature $T$.
2. **Generate** a random neighboring solution.
3. **Decide whether to accept** the new solution:
   - **Better solution** (lower loss): Always accept.
   - **Worse solution** (higher loss): Accept with probability:
$$
\boxed{
P(\text{accept})=
\begin{cases}
1 & \text{if } \Delta L\leq 0\\
e^{-\Delta L/T} & \text{if } \Delta L>0
\end{cases}
}
$$
where $\Delta L=L_{\text{new}}-L_{\text{current}}$.
4. **Decrease the temperature**, for example:
$$
T_{i+1}=rT_i,\qquad 0<r<1
$$
5. **Repeat** until the stopping condition is reached. Return the best solution found.

### **Exploration vs. Exploitation:**
- **High temperature:** More likely to accept worse solutions, encouraging exploration.
- **Low temperature:** Less likely to accept worse solutions, encouraging exploitation.

**Implementation:** Generate $\chi\sim\text{Uniform}(0,1)$ and accept the new solution if $\chi<P(\text{accept})$.

> [Visual example of Simulated Annealing](https://www.youtube.com/watch?v=iaq_Fpr4KZc)

````python
# Generate a random candidate solution near the current solution for n dimensions
def get_candidate_solution(current_solution, bounds, temp):
    return [np.clip(xi + np.random.normal() * (bounds[i][1] - bounds[i][0]) * temp,bounds[i][0],bounds[i][1]) for i, xi in enumerate(current_solution)]

# Simulated Annealing algorithm for n dimensions with different bounds
def simulated_annealing(objective, bounds, max_iterations, initial_temp, cooling_rate, dimensions):
    # Initialize with a random solution within the specific bounds for each dimension
    current_solution = [np.random.uniform(bounds[i][0], bounds[i][1]) for i in range(dimensions)]
    current_solution_cost = objective(current_solution)
    best_solution = current_solution[:]
    best_solution_cost = current_solution_cost

    # Current temperature
    temp = initial_temp

    for iteration in range(max_iterations):
        # Get a candidate solution respecting individual dimension bounds
        candidate_solution = get_candidate_solution(current_solution, bounds,temp)
        candidate_solution_cost = objective(candidate_solution)

        # Check if the candidate solution is better
        if candidate_solution_cost < current_solution_cost:
            current_solution, current_solution_cost = candidate_solution, candidate_solution_cost
            if candidate_solution_cost < best_solution_cost:
                best_solution, best_solution_cost = candidate_solution, candidate_solution_cost
        else:
            # Accept worse solutions with a certain probability
            if np.random.uniform(0, 1) < np.exp((current_solution_cost - candidate_solution_cost) / temp):
                current_solution, current_solution_cost = candidate_solution, candidate_solution_cost

        # Decrease the temperature
        temp *= cooling_rate

        # Optional: Output some information
        if (iteration+1) % 500 == 0:
            print(f"Iteration {iteration+1}, Temperature: {temp:.2f}, Current Solution: {current_solution}, Current Cost: {current_solution_cost}")

    return best_solution, best_solution_cost


# Define parameters
bounds = [(-1, 1), (-1, 1)]  # Individual bounds for each dimension
dimensions = len(bounds)  # Number of dimensions based on bounds
max_iterations = 2500  # Maximum number of iterations
initial_temp = 1  # Initial temperature
cooling_rate = 0.99  # Cooling rate

# Run the algorithm for n dimensions with different bounds
best_solution, best_solution_cost = simulated_annealing(
    griewank, bounds, max_iterations, initial_temp, cooling_rate, dimensions
)
print(f"Best solution in {dimensions}D with different bounds: {best_solution}, Best cost: {best_solution_cost}")
````

### Example of Simulated Annealing for any number of dimensions

````python

# Objective function for n-dimensional grid
def objective_function_nd(x):
    # Example objective function; replace with your specific function
    return sum(xi**2 for xi in x)

# Generate a random candidate solution near the current solution for n dimensions
def get_candidate_solution_nd(current_solution, bounds,temp):
    return [np.clip(xi + np.random.normal() * (bounds[i][1] - bounds[i][0]) * temp,bounds[i][0],bounds[i][1]) for i, xi in enumerate(current_solution)]

# Simulated Annealing algorithm for n dimensions with different bounds
def simulated_annealing(objective, bounds, max_iterations, initial_temp, cooling_rate, dimensions):
    # Initialize with a random solution within the specific bounds for each dimension
    current_solution = [np.random.uniform(bounds[i][0], bounds[i][1]) for i in range(dimensions)]
    current_solution_cost = objective(current_solution)
    best_solution = current_solution[:]
    best_solution_cost = current_solution_cost

    # Current temperature
    temp = initial_temp

    for iteration in range(max_iterations):
        # Get a candidate solution respecting individual dimension bounds
        candidate_solution = get_candidate_solution(current_solution, bounds,temp)
        candidate_solution_cost = objective(candidate_solution)

        # Check if the candidate solution is better
        if candidate_solution_cost < current_solution_cost:
            current_solution, current_solution_cost = candidate_solution, candidate_solution_cost
            if candidate_solution_cost < best_solution_cost:
                best_solution, best_solution_cost = candidate_solution, candidate_solution_cost
        else:
            # Accept worse solutions with a certain probability
            if np.random.uniform(0, 1) < np.exp((current_solution_cost - candidate_solution_cost) / temp):
                current_solution, current_solution_cost = candidate_solution, candidate_solution_cost

        # Decrease the temperature
        temp *= cooling_rate

        # Optional: Output some information
        if (iteration+1) % 500 == 0:
            print(f"Iteration {iteration+1}, Temperature: {temp:.2f}, Current Solution: {current_solution}, Current Cost: {current_solution_cost}")

    return best_solution, best_solution_cost


# Define parameters
bounds = [(-10, 10), (-20, 20), (-5, 5)]  # Individual bounds for each dimension
dimensions = len(bounds)  # Number of dimensions based on bounds
max_iterations = 2500  # Maximum number of iterations
initial_temp = 100  # Initial temperature
cooling_rate = 0.99  # Cooling rate

# Run the algorithm for n dimensions with different bounds
best_solution_nd, best_solution_cost_nd = simulated_annealing(
    objective_function_nd, bounds, max_iterations, initial_temp, cooling_rate, dimensions
)
print(f"Best solution in {dimensions}D with different bounds: {best_solution_nd}, Best cost: {best_solution_cost}")
````
