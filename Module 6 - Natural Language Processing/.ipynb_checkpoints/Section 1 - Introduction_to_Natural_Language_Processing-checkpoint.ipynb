## How Does NLP Work? From Text to Numerical Representations

**Core Idea:** NLP uses techniques to **extract information, represent text numerically, and learn patterns** for tasks such as classification, search, and text generation.

**Key Steps and Techniques:**
1. **Identify Patterns — Regular Expressions (Regex)**
    - Search for specific words or patterns in text.
        - Example: `\b[Cc]olleges?\b` matches "college", "College", and "colleges".
2. **Break Down Text — Tokenization & Preprocessing**
    - Split text into smaller units (tokens) and optionally clean or normalize it.
        - Example: `"I love machine learning!"` → `["I", "love", "machine", "learning", "!"]`

3. **Convert Text into Numbers — Bag of Words (BoW) & TF-IDF**
    - Represent documents using word frequencies or their importance.
        - Example: `"good wine"` → `[1, 1, 0]` for vocabulary `["good", "wine", "bad"]`.

4. **Represent Meaning — Word Embeddings**
    - Map words or tokens into dense numerical vectors that can capture semantic relationships.
        - Example: "king" and "queen" may have similar vector representations.

## I: Regular Expressions (Regex)



**Goal:** Identify specific words or text patterns using a search expression.

**Example: Detect "college" in Tweets**:
Suppose we want to label a tweet as **1** if it contains "college" or "colleges" (case-insensitive), and **0** otherwise.
<!-- - "I love The College of William & Mary!" → **1**
- "Many colleges offer scholarships." → **1**
- "I love my university!" → **0** -->

### **Regex Pattern:** 
`\bcolleges?\b` (with case-insensitive matching)
- `\b`: Word boundary (avoids matching "college" inside a longer word).
- `college`: Matches the exact text.
- `s?`: Matches zero or one `s` (college or colleges).
- **Case-insensitive:** Matches "college", "College", or "COLLEGE".

**Common Regex Patterns**:
| Pattern | Meaning | Example |
|---|---|---|
| `[tT]` | Lowercase or uppercase T | timber, Timber |
| `[A-Z]` | Any uppercase letter | A, B, Z |
| `[0-9]` | Any digit | 0, 5, 9 |
| `\w` | Word character (letters, digits, `_`) | a, 7, _ |
| `[^A-Z]` | Any character except an uppercase letter | a, 5, ! |

**Python Example:** `re.search(r'\bcolleges?\b', tweet, flags=re.IGNORECASE)`

**Reference:** [Regex Tutorial](https://www3.ntu.edu.sg/home/ehchua/programming/howto/Regexe.html)

## II: Tokenization: Breaking Text into Smaller Units

**Goal:** Split text into smaller units called **tokens**, which can be words, subwords, or characters.

**Original text:** `"This wine has a rich flavor!"` →
**Tokens:** `["This", "wine", "has", "a", "rich", "flavor", "!"]`

**Common Tokenization Approaches**
- **Word-level** (e.g., `"machine learning"` → `["machine", "learning"]`).
- **Subword-level** (e.g., `"unhappiness"` → `["un", "happi", "ness"]`, illustrative).
- **Character-level** (e.g., `"cat"` → `["c", "a", "t"]`).

**Special tokens** are often added for specific purposes, such as:
- <CLS>: Represents the start of a sequence (e.g., for classification tasks).
- <SEP>: Separates sequences (e.g., in question-answering tasks).
- <PAD>: Used for padding sequences to ensure uniform input lengths.

### **Why Tokenization?**
- **Classical NLP (BoW, TF-IDF):** Count or weight tokens to create numerical features.
- **Modern LLMs (ChatGPT):** Convert tokens into IDs and then dense embeddings for neural networks.
- **AI Pricing:** LLM APIs often charge based on the number of **input tokens** (prompts) and **output tokens** (generated responses), since processing more tokens generally requires more computation.

**Fun Fact:** 1 token is not necessarily 1 word! In English, 100 tokens correspond to roughly 75 words, although this varies by tokenizer and text.

## III: Bag of Words (BoW): Converting Text into Numerical Features

**Goal:** Represent a document as a vector based on **word occurrences or frequencies**, *ignoring* word order.

**Example: Two Movie Reviews**
- **Review 1:** "Jurassic World was amazing."
- **Review 2:** "Jurassic World was terrible."

BoW:
- **Step 1 — Tokenize and optionally remove stopwords** (e.g., "was").
- **Step 2 — Build a vocabulary:** `["jurassic", "world", "amazing", "terrible"]`
- **Step 3 — Count word occurrences:**
| Document | jurassic | world | amazing | terrible |
|---|---:|---:|---:|---:|
| Review 1 | 1 | 1 | 1 | 0 |
| Review 2 | 1 | 1 | 0 | 1 |

**Key Characteristics**
- **Simple and interpretable:** Each feature corresponds to a vocabulary word.
- **High-dimensional, sparse representations**: One feature per vocabulary word. This can increase computational costs and make it difficult to learn from rare words, although sparse storage can mitigate memory costs.
- **Limitations:** Ignores word order and contextual meaning; common words may dominate.

### Term Frequency–Inverse Document Frequency (TF-IDF)

**Goal:** Improve BoW by weighting words based on their importance in a document relative to the entire collection (**corpus**).

**Two Components**
1. **Term Frequency (TF):** How frequently does a word appear in a document?
$$
TF(t,d)=\frac{\text{Count of term }t\text{ in document }d}
{\text{Total number of terms in document }d}
$$
1. **Inverse Document Frequency (IDF):** How rare is the word across documents?
$$
IDF(t)=\log\left(\frac{N}{df(t)}\right)
$$
- where $N$ is the total number of documents and $df(t)$ is the number of documents containing term $t$.

**TF-IDF Score:**
$$
\boxed{TFIDF(t,d)=TF(t,d)\times IDF(t)}
$$
> Intuition: Gives more weight to words that distinguish documents.

### **TF-IDF Example: Movie Reviews**
- **"movie":** Appears in almost every review → Low IDF.
- **"masterpiece":** Appears in only a few reviews → High IDF.
- **"masterpiece" repeated in one review:** High TF and high IDF → High TF-IDF.

**Limitation:** Still produces sparse vectors and does not directly capture word order or contextual meaning.

*Note: These formulas illustrate a basic TF-IDF variant. Scikit-learn's `TfidfVectorizer` uses smoothed IDF and L2 normalization by default.*

## <font color='blue'>IV: Global Vectors for Word Representation (GloVe)</font>

**Reference:** [Stanford GloVe Project](https://github.com/stanfordnlp/glove)

**Key Idea:**

Unlike Bag-of-Words (BoW) and TF-IDF, which represent documents using word occurrence statistics, **GloVe learns dense word embeddings that capture semantic relationships between words**.

For example, word vectors can approximately capture analogies such as:

$$
\boxed{\vec{king}-\vec{man}+\vec{woman}\approx\vec{queen}}
$$

**How does GloVe learn these vectors?** By analyzing how frequently words appear together (**word co-occurrence**) across a large text corpus.

### Step 1: Construct a Co-occurrence Matrix

Given a **context window** (e.g., 10 words on each side), create a matrix $X$, where:

- $X_{ij}$ represents the weighted frequency with which word $j$ appears around word $i$.
- Words appearing in similar contexts tend to have similar meanings.

For example, *college* and *university* may frequently appear near words such as *students*, *professors*, and *education*.

### Step 2: Learn Word Vectors

GloVe assigns each word two trainable vectors: a word vector $\vec{w}_i$ and a context vector $\tilde{\vec{w}}_i$.

The goal is to approximate:
$$
\boxed{
\vec{w}_i^T\tilde{\vec{w}}_j+b_i+\tilde{b}_j
\approx \log(X_{ij})
}
$$

where:
- $\vec{w}_i,\tilde{\vec{w}}_j$: Dense vectors representing words and their contexts.
- $b_i,\tilde{b}_j$: Bias terms.
- $\log(X_{ij})$: Logarithm of the co-occurrence count.

**Intuition:** The dot product between two word vectors helps model how frequently the corresponding words appear together.

### Step 3: Minimize the Training Loss

GloVe learns these vectors by minimizing a weighted least-squares objective:

$$
\boxed{
J=\sum_{i,j:X_{ij}>0}f(X_{ij})
\left(
\vec{w}_i^T\tilde{\vec{w}}_j+b_i+\tilde{b}_j-\log(X_{ij})
\right)^2
}
$$
- The objective encourages **predicted co-occurrence relationships** to match **observed** ones.
- Weighting function to receive smaller weights for **rare co-occurrences** and cap weights for very common word. Common settings are $\alpha=0.75$ and $x_{\max}=100$.
$$
f(x)=
\begin{cases}
\left(\frac{x}{x_{\max}}\right)^\alpha, & x<x_{\max}\\
1, & x\geq x_{\max}
\end{cases}
$$

### Step 4: Use Pretrained GloVe Embeddings

**In practice, we usually use pretrained GloVe embeddings rather than training GloVe from scratch for each NLP problem.**

Stanford provides [pretrained GloVe vectors](https://github.com/stanfordnlp/glove) learned from large text corpora such as Wikipedia and Common Crawl.
For example:
- `glove.6B.50d`: 50-dimensional word vectors.
- `glove.6B.300d`: 300-dimensional word vectors.

**How do we use them for a specific task?**
1. **Download pretrained embeddings:** Each word is associated with a fixed-length vector.
2. **Convert text into vectors:** Look up the pretrained vector for each word in a document.
3. **Build document representations:** For example, average the word vectors or feed them into a neural network.
4. **Train a classifier:** Use the resulting representations for tasks such as sentiment analysis or spam detection.

For example, in tweet classification:
$$
\boxed{
\text{Tweet}\rightarrow\text{GloVe Vectors}\rightarrow\text{Classifier}\rightarrow\text{Label}
}
$$

>For specialized domains (e.g., medical or cybersecurity text), we may instead train **domain-specific** GloVe embeddings using a sufficiently large corpus.

### Takeaways

- **BoW / TF-IDF:** Represent documents using word occurrence statistics.
- **GloVe:** Learns compact, dense word vectors using global co-occurrence statistics.
- **Pretrained GloVe:** We can directly reuse existing word vectors without recomputing $X$ or retraining GloVe.
- **Advantage:** Word vectors can capture semantic similarity and certain relationships between words.
- **Limitation:** GloVe produces static embeddings, meaning a word has the same representation regardless of its context (e.g., *bank* in "river bank" versus "bank account").
- **Connection to LLMs:** Modern LLMs also use dense embeddings, but typically learn token embeddings during language-model training and generate context-dependent representations using Transformers.

## <font color='blue'>Extension: From Traditional NLP to Transformers and LLMs</font>

Traditional NLP methods such as **BoW, TF-IDF, and GloVe** convert text into numerical representations that machine learning models can process.

Modern NLP extends these ideas using **Transformers and Large Language Models (LLMs)**:
- **Attention Mechanism:** Helps models identify relationships between words, even when they are far apart in a sentence.
- **Contextual Representations:** Unlike GloVe's static embeddings, Transformers produce representations that depend on surrounding words.
    - Example: "bank" has different meanings in "river bank" and "bank account".
- **Large Language Models (LLMs):** Transformers can be trained on massive text datasets to predict tokens and generate coherent text.
    - Example: ChatGPT can answer questions, summarize documents, and generate text.

**Key Takeaway:** NLP has evolved from manually designed text features to **learned, context-aware representations** that support increasingly complex language tasks.

## Language Models & Perplexity

**Goal:** A language model predicts the probability of the **next token** given previous tokens.

### 1. How Does a Language Model Work?

For a sequence of tokens $w_1,w_2,\ldots,w_n$, the chain rule gives:

$$
P(w_1,\ldots,w_n)=\prod_{i=1}^{n}P(w_i\mid w_1,\ldots,w_{i-1})
$$

**Example:** Given `"The wine tastes"`, a language model might predict:

| Next token | Probability |
|---|---:|
| good | 0.50 |
| sweet | 0.30 |
| terrible | 0.10 |
| Other | 0.10 |

**Classical n-gram models** estimate probabilities from word frequencies; **modern LLMs** learn these probabilities using neural networks.

### 2. Perplexity: How Well Does a Language Model Predict?

**Perplexity (PPL)** measures how surprised a language model is by the actual tokens in a test sequence.

$$
\boxed{\text{PPL}=\exp\left(-\frac{1}{n}\sum_{i=1}^{n}\log P(w_i\mid w_{<i})\right)}
$$

**Simple Example:** Suppose a model assigns probability $0.25$ to each of 3 actual next tokens:

$$
\text{PPL}=(0.25\times0.25\times0.25)^{-1/3}=\boxed{4}
$$

- **Lower PPL = Better predictions** (on the same test data and tokenization).
- **PPL = 1:** The model assigns 100% probability to every actual next token.
- **PPL = 4:** Roughly equivalent to choosing among 4 equally likely possibilities at each step.

### 3. Connection to Modern LLMs

- **Training:** Minimize cross-entropy loss, which also minimizes perplexity.
- **Evaluation:** Use perplexity to measure next-token prediction performance.
- **Limitation:** Low perplexity does not necessarily mean responses are accurate, useful, or factually correct.

**Key Takeaway:** Language models predict tokens using probabilities; **perplexity measures how well they predict the actual sequence**.

## Faster Preprocessing

Fast text preprocessing function that handles:

- Lowercasing
- Punctuation removal
- Stopword removal
- Lemmatization (optional -- input adjustable choice)

Arguments:

    - texts: List of strings or single string to process
    - stop_words: Set of stopwords (if None, uses NLTK English stopwords)
    - lemmatize: Whether to perform lemmatization

Returns:
    Processed texts (list or string, matching input type)

Handle single string input

````python
def fast_preprocess(texts, lemmatize=True):

    single_input = isinstance(texts, str)
    if single_input:
        texts = [texts]



    stop_words = set(stopwords.words('english'))

    # Create translator for punctuation removal (faster than regex for this task)
    translator = str.maketrans('', '', string.punctuation)

    # Initialize lemmatizer if needed
    lemmatizer = WordNetLemmatizer() if lemmatize else None

    processed_texts = []
    for text in texts:
        # Lowercase
        text = text.lower()

        # Remove punctuation
        text = text.translate(translator)

        # Tokenize (needed for both stopword removal and lemmatization)
        tokens = text.split()

        # Remove stopwords
        tokens = [token for token in tokens if token not in stop_words]

        # Lemmatize if requested
        if lemmatize:
            tokens = [lemmatizer.lemmatize(token) for token in tokens]

        # Rejoin into text
        processed_texts.append(' '.join(tokens))

    # Return in same format as input
    return processed_texts[0] if single_input else processed_texts
````

````python
allreviews = fast_preprocess(reviews)
````

````python
# the first example uses only 1000 observations
wine_data = pd.read_csv('https://github.com/dvasiliu/AML/blob/main/Data%20Sets/winemagdata130kv2.csv?raw=true',quoting=2)
wines = wine_data[["description","points"]]
wines_subset = wines.sample(1000,random_state=1234).reset_index(drop=True)
corpus = []
````

````python
reviews = wines_subset.description.values #
````

````python
def fast_remove_stopwords(texts):

    stop_words = set(stopwords.words('english'))

    # Compile regex for efficiency (matches whole words only)
    pattern = r'\b(?:{})\b'.format('|'.join(stop_words))
    regex = re.compile(pattern, re.IGNORECASE)

    # Process all texts with a single function call
    return [regex.sub('', text) for text in texts]
````

````python
def fast_preprocess(texts, lemmatize=True):

    # Fast text preprocessing function that handles:
    # - Lowercasing
    # - Punctuation removal
    # - Stopword removal
    # - Lemmatization (optional)

    # Arguments:
    #     texts: List of strings or single string to process
    #     stop_words: Set of stopwords (if None, uses NLTK English stopwords)
    #     lemmatize: Whether to perform lemmatization

    # Returns:
    #     Processed texts (list or string, matching input type)

    # Handle single string input
    single_input = isinstance(texts, str)
    if single_input:
        texts = [texts]



    stop_words = set(stopwords.words('english'))

    # Create translator for punctuation removal (faster than regex for this task)
    translator = str.maketrans('', '', string.punctuation)

    # Initialize lemmatizer if needed
    lemmatizer = WordNetLemmatizer() if lemmatize else None

    processed_texts = []
    for text in texts:
        # Lowercase
        text = text.lower()

        # Remove punctuation
        text = text.translate(translator)

        # Tokenize (needed for both stopword removal and lemmatization)
        tokens = text.split()

        # Remove stopwords
        tokens = [token for token in tokens if token not in stop_words]

        # Lemmatize if requested
        if lemmatize:
            tokens = [lemmatizer.lemmatize(token) for token in tokens]

        # Rejoin into text
        processed_texts.append(' '.join(tokens))

    # Return in same format as input
    return processed_texts[0] if single_input else processed_texts
````
