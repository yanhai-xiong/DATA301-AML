## <font color='deepskyblue'>**Example 4**</font> A Fibonacci sequence.

Iterative approach to generating the Fibonacci sequence, i.e. the sequence $(f_n)$ of integers such that $$f_0=0,\,\, f_1=1\,\,\, \text{and}\,\, f_{n+1}=f_n+f_{n-1}$$ for all $n\geq 1.$

````python
class Fib2:
  def __init__(self,a=0,b=1):
    self.a = a
    self.b = b
    self.max = 20
  def __iter__(self):
      return self
  def __next__(self):
    # Return the next item or raise StopIteration.
    if self.b < self.max:
        self.a, self.b = self.b, self.a + self.b
        return self.b
    else:
        raise StopIteration
````

````python
myfib2 = Fib2(a=0.2,b=1.5)
for item in myfib2:
    print(item)
````

## Example Collatz Conjecture

Take any number and if it's even divide it by 2, and if it's odd, multiply it by 3 and add 1. Then repeat.
Define the piecewise function f, you are going to compute f(f(f(f...(n_times)))

````python
# example with classes
class collatz:
    # attributes initialization
    def __init__(self,x=1):
        self.x = x
    # method definition
    def f(self):
        if self.x%2 == 0:
            return self.x/2
        else:
            return 3*self.x+1
    def __iter__(self):
        return self
    def __next__(self):
        self.x = self.f()
        return self.x
````

````python
myclass = collatz(x=233471)
# i is a counter variable
i = 1
# next(myclass) is a built-in Python function that manually 
# triggers the __next__ method inside your class
while next(myclass)>1:
    i += 1
print('The Collatz conjecture worked in: '+str(i),' iterations.')
````

````python
myclass = collatz(x=233471)
# i is a counter variable
i = 1
while next(myclass)>1:
    # we would get the ith iteration result here
    if i == 10:
        print("Value of x at the end of 10th iter:", myclass.x)
    i += 1
print('The Collatz conjecture worked in: '+str(i),' iterations.')
````

### Debug
Below is a submission in class:

```python

```

1. When you use a for loop `for x in myclass` on an iterator, Python automatically calls next() behind the scenes to assign a value to x.
1. `print(next(myclass))` advances it again since `next(myclass)` is a built-in Python function that manually triggers the `__next__` method inside your class
1. Note that `print(myclass)` would not work since there is no `__str__` function defined, i.e., myclass doesn't have a str representation format for printing 
1. Your collatz class never raises a StopIteration exception, so there would be infinite loops.

````python
# here is how to fix it 
myclass = collatz(x=233471)
iter_ = 1
for x in myclass:
    print(myclass.x)
    iter_ += 1
    if iter_ == 11:
        break
````

````python
# here is how to fix it (another option)
# enumerate tracks the index (0, 1, 2...) automatically
myclass = collatz(x=233471)
for index, x in enumerate(myclass):
    if index == 10:
        break  # Stops the loop after 10 outputs
    print(x)
````
