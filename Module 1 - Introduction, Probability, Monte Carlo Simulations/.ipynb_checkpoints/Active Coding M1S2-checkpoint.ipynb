```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

## Solve the Monty Hall Problem via Simulations

```python
# this is just a list but is not a vector or array
doors = ['goat', 'goat', 'car']
```

```python
# how to randomly pick one of the doors?
np.random.shuffle(doors)
doors[0]
```

```python
# how frequently are you winning the car if you stick with your original choice?
stick_wins = 0
switch_wins = 0
n = 10000
for _ in range(n):
    np.random.shuffle(doors)

    # Player initially chooses door 0, so
    available_to_open = [
        i for i in [1, 2]
        if doors[i] == 'goat'
    ]

    # this simulated the operator's action 
    opened = np.random.choice(available_to_open) 

    # The remaining unopened door
    switch_door = next(
        i for i in [1, 2]
        if i != opened
    )

    if doors[0] == 'car':
        stick_wins += 1

    if doors[switch_door] == 'car':
        switch_wins += 1
        
print("The experimental probability for winning by stick to the original choice is "+str(stick_wins/n))
print("The experimental probability for winning by switching the original choice is "+str(switch_wins/n))
```

```python
# here we program the actual Monte Carlo Simulation
doors = ["goat","goat","car"]
n = 10000
# approximated results
switch_win_probability = []
stick_win_probability = []

plt.axhline(y=0.666, color='red', linestyle='--')
plt.axhline(y=0.333, color='green', linestyle='--')

def monte_carlo(n):

  switch_wins = 0
  stick_wins = 0

  for i in range(n):
     random.shuffle(doors)
     # in k we have the player's choice
     k = random.randrange(3)

     if doors[k] != 'car':
       switch_wins +=1

     else:
       stick_wins +=1

     switch_win_probability.append(switch_wins/(i+1))
     stick_win_probability.append(stick_wins/(i+1))

  plt.plot(switch_win_probability,label='SwitchWin Probability')
  plt.plot(stick_win_probability, label='StickWin Probability')
  plt.tick_params(axis='x', colors='navy')
  plt.tick_params(axis='y', colors='navy')
  plt.xlabel('Repetitions of Experiment',fontsize=14,color='green')
  plt.ylabel('Experimental Probability',fontsize=14,color='green')
  plt.legend()

  print('Winning probability if you always switch:', switch_win_probability[-1])
  print('Winning probability if you always stick to your original choice:', stick_win_probability[-1])

monte_carlo(5000)
```

## Use Monte Carlo Simulations to Estimate the irrational number $\pi$

```python
# the main idea is to think of a circle of radius 1 inscribed in a square of side of length 2
# We are going to throw darts at the square and count how many land also inside the circle
hits_inside_circle = 0
n = 10000000
for _ in range(n):
    x = np.random.uniform(-1,1)
    y = np.random.uniform(-1,1)
    if x**2+y**2<=1:
        hits_inside_circle += 1
print('The MC approximation of pi is :'+str(4*hits_inside_circle/n))
```

```python
pi_values = []

def monte_carlo(n):
  in_circle = 0
  out_circle = 0
  for j in range(n):
    x = random.randrange(250,500)
    y = random.randrange(120,370)

    # check if the number lies outside the circle
    if (x-375)**2+(y-245)**2>125**2:
      out_circle = out_circle + 1
    else:
      in_circle = in_circle + 1

    approx_pi = 4.0 * in_circle / (in_circle + out_circle)
    pi_values.append(approx_pi)

  return pi_values

pi_values = monte_carlo(10000)
plt.axhline(y=np.pi,color='red',linestyle='--')
plt.plot(pi_values)
plt.tick_params(axis='x', colors='navy')
plt.tick_params(axis='y', colors='navy')
plt.xlabel('Iteration',fontsize=14,color='green')
plt.ylabel('Probability',fontsize=14,color='green')

answer = pi_values[-1]
print(answer)
```
