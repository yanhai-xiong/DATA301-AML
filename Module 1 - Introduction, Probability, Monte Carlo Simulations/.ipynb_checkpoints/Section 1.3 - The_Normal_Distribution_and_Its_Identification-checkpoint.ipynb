## <font color='blue'> The Law of Probable Errors </font>

The story began when in 1801 an italian astonomer, Giuseppe Piazzi, discovered a new object in our solar system, the dwarf planet (asteroid) Ceres:

https://www.jpl.nasa.gov/news/ceres-keeping-well-guarded-secrets-for-215-years

A german mathematician, Carl Friedrich Gauss, helped relocate the position of Ceres and confirmed the discovery.

"... *for it is now clearly shown that the orbit of a heavenly body may be determined quite nearly from good observations embracing only a few days; and this without any hypothetical assumption*." - Carl Friedrich Gauss

### Gauss proposed the *law of probable errors:*

1. Small errors are more likely than large errors;
2. The likelihood of errors of the same magnitude but different signs, such as  `x` and `-x`, are equal (the distribution is symmetrical);
3. When several measurements are taken of the same quantity, the average (arithmetic mean) is the most likely value.

### Bell-Shaped Density Curve
A normal distribution has a bell-shaped density curve described by its mean `μ` and standard deviation `σ`. The density curve is `symmetrical`, centered about its mean, with its spread determined by its standard deviation showing that data near the mean are more frequent in occurrence than data far from the mean.
A normal distribution is a continuous probability distribution defined by its probability density function (PDF):

$$\large
f(x \mid \mu, \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\!\left(-\frac{(x-\mu)^2}{2\sigma^2}\right),
$$

where:
- $ \mu $ is the **mean** (or expectation) of the distribution.
- $ \sigma^2 $ is the **variance**, and $ \sigma $ is the standard deviation.
- $ \exp(\cdot) $ denotes the exponential function.

### **Probability Interpretation**: 
The probability that the random variable $ X $ falls within a particular range $[a, b]$ is given by the integral of the PDF over that range:
   $$\large
   P(a \leq X \leq b) = \int_{a}^{b} f(x) \, dx
   $$

That is to say the area under the graph of the density function when $X$ is between $a$ and $b.$ The integral sign means **accumulation** or **sum** of all frequencies over a given interval.

### Key Characteristics

- **Symmetry:** The normal distribution is symmetric about its mean $ \mu $. This implies that the probability of a value being above the mean is the same as the probability of it being below the mean.
- **Bell Shape:** The familiar bell shape arises because values closer to the mean are more likely, and the probability decreases exponentially as you move away from the mean.
- **Empirical Rule (68-95-99.7 Rule):** In a normal distribution:
  - About **68%** of the data falls within one standard deviation ($ \mu \pm \sigma $).
  - About **95%** falls within two standard deviations ($ \mu \pm 2\sigma $).
  - About **99.7%** falls within three standard deviations ($ \mu \pm 3\sigma $).
- **Total Area:** The area under the PDF over the entire real line is 1, which is a requirement for any probability density function.

### The Standard Normal Distribution

A special case of the normal distribution is the **standard normal distribution**, which has a mean of 0 and a variance of 1. Its PDF is given by:

$$
\phi(z) = \frac{1}{\sqrt{2\pi}} \exp\!\left(-\frac{z^2}{2}\right),
$$

where $ z $ is the standardized variable.

### Standardization

Any normal random variable $ X \sim N(\mu, \sigma^2) $ can be converted to a standard normal variable $ Z $ using the transformation:

$$
Z = \frac{X - \mu}{\sigma}.
$$

This process, known as **standardization**, allows us to use standard normal tables (or software) to compute probabilities.

## <font color='blue'> What is a Normality Test? </font>

> **Main Idea**: We compare the `empirical density function of the data` of the data with the `theoretical density of a standard normal distribution` of a normal distribution. If they have an approximately `linear relationship`, the data are consistent with a normal distribution.

Assume we have a univariate set of values ( we have a sample from a univariate distribution.) We are checking, based on some calculations with the sample data, if the univariate distribution is a normal one.

### **Why it's important**: 
It checks if your data fits a standard bell-shaped curve, which helps you *choose the correct statistical tool* and *trust your final results*. 
Many common statistical methods—like t-tests, ANOVA, and linear regression—are parametric tests. They assume your data is normally distributed.

### **General approach**: 
Compare the cumulative distribution functions (integral of PDF).
$ \large CDF(x):= P(X<x) = \int_{-∞}^{x}f(t)dt$

We want to recast this matching procedure onto the backdrop of a linear correlation situation; this means we want to compare the two cumulative distribution functions. To explain, we want the empirical percentiles to correlate linearly with the theoretical percentile of a standard normal distribution.

<p class="fragment-fade">The concept is useful in applications where we want to check if a random variable follows a certain distribution. <p>

**IMPORTANT**: In most cases we standardize the values of the random variable, e.g we compute z-scores.

The linear correlation coefficient is defined (Pearson, Karl (20 June 1895). "Notes on regression and inheritance in the case of two parents". Proceedings of the Royal Society of London. 58: 240–242.)

### Pearson Correlation Coefficient
$$\large r: = \frac{1}{n-1}\sum_{i=1}^{n} \left(\frac{x_i - \bar{x}}{s_x}\right)\left(\frac{y_i - \bar{y}}{s_y}\right)$$

Here $\bar{x}$ is the mean of $x$, $\bar{y}$ is the mean of $y$ and, $s_x$ is the standard deviation of $x$ and $s_y$ is the standard deviation of $y.$

> (Strong negative linear relationship) -1 < r < 1 (Strong positive linear relationship);\
> r≈0 Weak/no linear relationship

### Test statistic for the correlation coefficient:

$$\large t=\frac{r\sqrt{n-2}}{\sqrt{1-r^2}}$$
where $n$ is the number of observations and $r$ represents the correlation coefficient computed from the data.

> The test statistic converts the observed correlation `r` into a standardized [measure](https://www.itl.nist.gov/div898/handbook/eda/section3/eda3672.htm) of evidence **against** the null hypothesis that there is no linear correlation between `X` and `Y`, while accounting for the sample size.

- $abs(t)$ small ⇒ weak evidence against null hypothesis
- $abs(t)$ large ⇒ strong evidence against null hypothesis


## <font color='blue'> Normality Test 1: The Kolmogorov-Smirnov Test </font>

The test (named after two Soviet mathematicians) is defined as:
- **H0 (the null hypothesis):**	The data follow a specified distribution.
- **H1 (the alternative hypothesis):** The data do not follow the specified distribution

The main idea is that we focus on how much the empirical cumulative distribution function is different from the theoretical cumulative distribution function, and we may consider:
$$\large \sup_{x} |ECDF(x) - CDF(x)|$$

where $ECDF(x)$ means the emprirical cumulative distribution function:
$$\large ECDF(x):= \frac{1}{n}\sum \mathbb{1}(t)_{t<x}$$

and, $CDF$ stands for the cumulative distribution function:
$$ \large CDF(x):= \int_{-\infty}^{x}f(t)dt.$$  

### $f$ The Probability Density Function.

If we order the observations, such as $x_i\leq x_j$ when $i\leq j$, then the test statistic is formally defined by:

$$\large D:=\max_{1\leq i\leq n}\left\{CDF(x_i)-\frac{i-1}{n},\frac{i}{n}-CDF(x_i)\right\}$$

The mathematical notation means that we add $1$ for each $t$ less than $x$ and $n$ represents the sample size.

## <font color='blue'> Normality Test 2: The Anderson-Darling Test </font>

The test is defined as:
- **H0 (the null hypothesis)**:	The data follow a specified distribution.
- **H1 (the alternative hypothesis)**: The data do not follow the specified distribution

The test statistic is defined as:
$$\large AD := -n - \sum_{i=1}^{n} \frac{2i-1}{n}\left[\ln(CDF(x_i))+\ln(1-CDF(x_{n+1-i})\right] $$

Then we compare $AD^2$ with critical values to decide. The critical values for the Anderson-Darling test are dependent on the specific distribution that is being tested.

## <font color='blue'> Normality Test 3: The Shapiro-Wilk Test </font>

The **Shapiro-Wilk test** is a widely used statistical test that helps determine whether a given sample comes from a normally distributed population. Introduced by Samuel Shapiro and Martin Wilk in 1965, this test is especially popular for its power in detecting departures from normality, even with small sample sizes. Below is a comprehensive introduction to the Shapiro-Wilk test covering its purpose, methodology, computation, interpretation, assumptions, and limitations.

### Main Idea
The primary goal of the Shapiro-Wilk test is to assess the null hypothesis that a sample $ \{x_1, x_2, \dots, x_n\} $ is drawn from a normally distributed population. It is particularly sensitive in detecting both skewness and kurtosis, making it a robust choice for normality testing.

### Hypotheses
- **Null Hypothesis ($H_0$)**: The data are drawn from a normal distribution.
- **Alternative Hypothesis ($H_1$)**: The data are not drawn from a normal distribution.

### The Test Statistic
The Shapiro-Wilk test statistic, denoted as $ W $, is defined as:
$$
W = \frac{\left(\sum_{i=1}^{n} a_i \, x_{(i)}\right)^2}{\sum_{i=1}^{n} \left(x_i - \bar{x}\right)^2}
$$

where:
- $ x_{(i)} $ are the ordered sample values (i.e., $ x_{(1)} \leq x_{(2)} \leq \cdots \leq x_{(n)} $).
- $ \bar{x} $ is the sample mean.
- $ a_i $ are constants computed from the expected values and variances of the order statistics of a sample from the standard normal distribution. These coefficients are derived so that the numerator represents a measure of how well the ordered sample values match the expected order under normality.

### Interpretation of $ W $
$$
W = \frac{\left(\sum_{i=1}^{n} a_i \, x_{(i)}\right)^2}{\sum_{i=1}^{n} \left(x_i - \bar{x}\right)^2}
$$

- **$ W $ near 1**: Indicates that the sample data are consistent with a normal distribution.
- **$ W $ significantly less than 1**: Suggests deviations from normality.

A smaller value of $ W $ implies that the data are less likely to have come from a normal distribution.

### Computational Aspects
The coefficients $ a_i $ are precomputed based on the sample size $ n $ and the covariance matrix of the order statistics of a standard normal distribution. Most statistical software packages (such as R, Python’s SciPy, SAS, and SPSS) have built-in functions to compute these coefficients and the test statistic, so manual calculation is rarely necessary.

### P-value Computation
Once the test statistic $ W $ is calculated, a corresponding p-value $p=P(W ≤ W_{obs} ∣H0)$ is determined. This p-value is the *probability* of observing a test statistic as extreme as, or more extreme than, the computed $ W $ under the null hypothesis of normality. If the p-value is less than a pre-specified significance level (commonly $ \alpha = 0.05 $), the null hypothesis is rejected.

### Practical Application and Interpretation
- **Step 1:** Collect your data sample.
- **Step 2:** Compute or use software to determine the Shapiro-Wilk test statistic $ W $ and the associated p-value.
- **Step 3:** Compare the p-value with your chosen significance level:
  - **If $ p \geq \alpha $:** Do not reject $ H_0 $; there is no strong evidence against normality.
  - **If $ p < \alpha $:** Reject $ H_0 $; the data provide evidence of non-normality.

#### Code Example
For example, in **Python** using the `scipy.stats` module:

```python
import scipy.stats as stats

# Example data sample
data = [your data here]

# Perform the Shapiro-Wilk test
W, p_value = stats.shapiro(data)

print("Shapiro-Wilk Test Statistic:", W)
print("P-value:", p_value)
```

### Assumptions
- **Independence:** The observations in the sample should be independent.
- **Continuous Data:** The test is designed for continuous data. While it can be applied to discrete data, caution should be exercised in interpretation.
- **Sample Size:** The Shapiro-Wilk test is most effective for small to moderate sample sizes (commonly recommended for $ n \leq 2000 $). For very large samples, even small deviations from normality may lead to rejection of the null hypothesis.

### Limitations
- **Sensitivity:** With large sample sizes, the test may detect trivial deviations from normality that are not practically significant.
- **Power Against Certain Alternatives:** While powerful against many departures from normality, no test is universally powerful against all forms of non-normality.

````python
# useful class for the visualization of frequencies
class distribution_plot:
  def __init__(self,dist,bins=11,hist_color='lightblue',pdf_color='red',fill_region=False,fill_color='lightgreen',fill_bounds=[-1,1],edge_color='k',lw_pdf=2):
    self.dist = dist
    self.bins = bins
    self.hist_color = hist_color
    self.pdf_color = pdf_color
    self.edgecolor = edge_color
    self.lw_pdf = lw_pdf
    self.fill_region = fill_region
    self.fill_color = fill_color
    self.fill_bounds = fill_bounds


  def fit(self,data):
    fig, ax = plt.subplots(figsize=(6, 5))
    # Create the histogram
    sns.histplot(data,
                bins=self.bins,
                kde=False,
                color=self.hist_color,
                alpha=0.5,
                edgecolor=self.edgecolor,
                ax=ax)

    # Calculate the parameters for the normal distribution fit
    params = self.dist.fit(data)

    # Create the x values for the normal distribution curve
    xmin, xmax = ax.get_xlim()
    x = np.linspace(xmin, xmax, 100)

    # Calculate the y values for the normal distribution curve
    p = self.dist.pdf(x, *params)

    # Plot the normal distribution curve, scaling it to the histogram
    # We need to scale it because histplot uses counts, while norm.pdf is a density
    scale_factor = len(data) * (xmax - xmin) / self.bins # Approximate bin width
    ax.plot(x, p * scale_factor, 'r', linewidth=self.lw_pdf)

    ax.set_xlabel('Random Variable', fontsize=16)
    ax.set_ylabel('Frequency', fontsize=16)
    ax.set_title('Data Distribution', fontsize=16)
    if self.fill_region:
      l = ax.lines[0]
      xx = l.get_xydata()[:,0]
      yy = l.get_xydata()[:,1]
      ax.fill_between(xx,yy, where=(self.fill_bounds[0]<=xx) & (xx <= self.fill_bounds[1]), color=self.fill_color,alpha=0.5)
      plt.savefig('example_distribution.png',dpi=300)
    plt.show()
````

````python
def test_ks(x):
  n = len(x)
  x_sorted = np.sort(x)
  d = 0.0
  for i in range(n):
    if d < np.max([CDF(x_sorted[i])-i/n,(i+1)/n-CDF(x_sorted[i])]):
      d = np.max([CDF(x_sorted[i])-i/n,(i+1)/n-CDF(x_sorted[i])])
  return d
````

````python
test_ks(zscore(x))
````

````python
# this shows how we can generate a table of critical values for the K-S test

from scipy.stats import ksone

def ks_critical_value(n_trials, alpha):
    return ksone.ppf(1-alpha/2, n_trials)

trials = range(1, 121)
alphas = [0.1, 0.05, 0.02, 0.01]

# Print table headers
print('{:<6}|{:<6} Level of significance, alpha'.format(' ', ' '))
print('{:<6}|{:>8} {:>8} {:>8} {:>8}'.format(*['Trials'] + alphas))
print('-' * 42)
# Print critical values for each n_trials x alpha combination
for t in trials:
    print('{:6d}|{:>8.5f} {:>8.5f} {:>8.5f} {:>8.5f}'
          .format(*[t] + [ks_critical_value(t, a) for a in alphas]))
    if t % 10 == 0:
        print()
````
